### **Descriptive Analysis** 

#### **Categorical Data**

To explore categorical attributes use <code>matplotlib.pyplot.subplots</code>. Create a grid containing three bar charts to visualise the distribution of data across categories. 

**Categorical Columns**
* Sex
* Region
* Smoker

Use the variable axs to plot multiple Axes. Access plotting capabilities by calling the index for the axs required. For example to countplot 'sex' assign  <code>ax=axs[0]</code> or to countplot 'region' assign <code>ax=axs[1]</code> and to countplot 'smoker' assign <code>ax=axs[2]</code>.

Set theme using Seaborn library to <code>'darkgrid'</code>. This feature enhances the visualisation by adding a grid on the charts with a colour tint for the results to pop. 

NOTE: CoPilot helped fix a ValueError by adding missing <code>style=</code> inside the parentheses to apply theme.

```python
# Set design theme
sns.set_theme(style="darkgrid")

# Create a grid for 1 row, 3 columns
fig, axs = plt.subplots(1, 3, figsize=(15, 5)) 

# Count distribution for sex category 
sns.countplot(data=df_insurance, x="sex", ax=axs[0])
axs[0].set_title("Sex Distribution") # Set chart title 
axs[0].set_xlabel("Sex") # Title X axis label as sex  
axs[0].set_ylabel("Count") # Title Y axis label to indicate count  

# Count distribution for region category
sns.countplot(data=df_insurance, x="region", ax=axs[1])
axs[1].set_title("Regional Distribution") # Set chart title 
axs[1].set_xlabel("Region") # Title X axis label as region  
axs[1].set_ylabel("Count") # Title Y axis label to indicate count  

# Count distribution for smoker category 
sns.countplot(data=df_insurance, x="smoker", ax=axs[2])
axs[2].set_title("Smoker vs Non-Smoker Distribution") # Set chart title 
axs[2].set_xlabel("Smoker") # Title X axis label as smoker  
axs[2].set_ylabel("Count") # Title Y axis label to indicate count  

# Print grid visualising distribution counts 
plt.show()
```

Barplots were used to represent the distribution of categorical counts because the spread can explictly present the balance. For example 'Smoker vs Non-Smoker Distribution' shows the distribution is uneven. Thus it is important to consider this when considering smoker-status charges.

<br/>

#### **Numerical Distribution**

To explore numerical data distribution boxplots will be used to forsee the balance in each category. 

**Numerical Columns**
* Age
* BMI
* Children
* Charges


To simplify subplot process use NumPy method <code>axs = axs.flatten()</code> to transform 2d arrary into a 1d array.

```python
# Create a grid for 2 rows, 2 columns
fig, axs = plt.subplots(2, 2, figsize=(14, 10))

# Flatten axes into 1d array 
axs = axs.flatten()

# Boxplot distribution for age
sns.boxplot(data=df_insurance, y="age", ax=axs[0]) 
axs[0].set_title("Age Distribution") # Set boxplot title 
axs[0].set_ylabel("Age") # Title Y axis label as age

# Boxplot distribution for bmi
sns.boxplot(data=df_insurance, y="bmi", ax=axs[1])
axs[1].set_title("BMI Distribution") # Set boxplot title 
axs[1].set_ylabel("BMI") # Title Y axis label as bmi

# Boxplot distribution for number of children
sns.boxplot(data=df_insurance, y="children", ax=axs[2])
axs[2].set_title("Number of Children") # Set boxplot title
axs[2].set_ylabel("Children Count") # Title Y axis label as children count

# Boxplot distribution for charges 
sns.boxplot(data=df_insurance, y="charges", ax=axs[3])
axs[3].set_title("Insurance Charges") # Set boxplot title
axs[3].set_ylabel("Charges") # Title Y axis label as charges

# Print grid visualising distribution counts
plt.show()
```

### **Correlation Analysis** 

A correlation analysis can present connections between different attributes present in the dataset. To visualise the correlation between smoking status and age or number of children on insurance charges use Seaborn to create a scattergraph that includes to represent the categorical attribute as a hue.

<br/>

#### **Smoking Status, Age and Charges**

In the hypothesis a question is raised on the correlation between age, smoking status and charges. Seaborn provides a scatterplot including a hue that can represent a categorical attribute combined with the numerical datatypes age and charges. 

```python
# Create scatterplot with seaborn 
sns.scatterplot(
    data=df_insurance,
    x="age", # Asign X axis to age 
    y="charges", # Assign Y axis to charges
    hue="smoker" # Assign hue to smoker 
)

plt.title("Correlation between Age and Smoking Status on Charges") # Set scatterplot title 
plt.xlabel("Age") # Title X axis label as age
plt.ylabel("Insurance Charges") # Title Y axis label as insurance charges 

# Print scatterplot
plt.show()
```

The scatterplot shows there is a correlation between age, smoking and charges. Smokers are charged almost double the amount than non-smokers over a lifetime.
