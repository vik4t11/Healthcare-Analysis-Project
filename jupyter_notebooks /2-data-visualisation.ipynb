### **Descriptive Analysis** 

In the descriptive analysis, distribution is primarily explored. Grids are created to visualise the distribution across categorical and numerical columns.

<br/>

#### **Categorical Data**

To explore categorical attributes, use <code>matplotlib.pyplot.subplots</code>. Create a grid containing three bar charts to visualise the distribution of data across the categories.

**Categorical Columns**

* Sex
* Region
* Smoker

Use the variable <code>axs</code> to plot multiple Axes. Access each plotting area by calling the required index from <code>axs</code>. For example, to create a countplot for 'sex', assign <code>ax=axs[0]</code>; for 'region', assign <code>ax=axs[1]</code>; and for 'smoker', assign <code>ax=axs[2]</code>.

Set the theme using the Seaborn library to <code>'darkgrid'</code>. This feature enhances the visualisation by adding a grid to the charts with a colour tint to make the results stand out.

NOTE: CoPilot helped fix a ValueError by adding the missing <code>style=</code> inside the parentheses to apply the theme.

```python
# Set design theme
sns.set_theme(style="darkgrid")

# Create a grid for 1 row, 3 columns
fig, axs = plt.subplots(1, 3, figsize=(15, 5)) 

# Count distribution for sex category 
sns.countplot(data=df_insurance, x="sex", ax=axs[0])
axs[0].set_title("Sex Distribution") # Set chart title 
axs[0].set_xlabel("Sex") # Title X axis label as sex  
axs[0].set_ylabel("Count") # Title Y axis label to indicate count  

# Count distribution for region category
sns.countplot(data=df_insurance, x="region", ax=axs[1])
axs[1].set_title("Regional Distribution") # Set chart title 
axs[1].set_xlabel("Region") # Title X axis label as region  
axs[1].set_ylabel("Count") # Title Y axis label to indicate count  

# Count distribution for smoker category 
sns.countplot(data=df_insurance, x="smoker", ax=axs[2])
axs[2].set_title("Smoker vs Non-Smoker Distribution") # Set chart title 
axs[2].set_xlabel("Smoker") # Title X axis label as smoker  
axs[2].set_ylabel("Count") # Title Y axis label to indicate count  

# Print grid visualising distribution counts 
plt.show()
```

Barplots were used to represent the distribution of categorical counts because the spread can explicitly present the balance across categories. For example, the 'Smoker vs Non-Smoker Distribution' shows that the distribution is uneven. Therefore, it is important to consider this imbalance when analysing charges in relation to smoker status.

<br/>

#### **Numerical Distribution**

To explore the distribution of numerical data, boxplots will be used to assess the spread and balance of each attribute.

**Numerical Columns**
* Age
* BMI
* Children
* Charges


To simplify the subplot process, use the NumPy method <code>axs = axs.flatten()</code> to transform the 2D array into a 1D array.

```python
# Create a grid for 2 rows, 2 columns
fig, axs = plt.subplots(2, 2, figsize=(14, 10))

# Flatten axes into 1d array 
axs = axs.flatten()

# Boxplot distribution for age
sns.boxplot(data=df_insurance, y="age", ax=axs[0]) 
axs[0].set_title("Age Distribution") # Set boxplot title 
axs[0].set_ylabel("Age") # Title Y axis label as age

# Boxplot distribution for bmi
sns.boxplot(data=df_insurance, y="bmi", ax=axs[1])
axs[1].set_title("BMI Distribution") # Set boxplot title 
axs[1].set_ylabel("BMI") # Title Y axis label as bmi

# Boxplot distribution for number of children
sns.boxplot(data=df_insurance, y="children", ax=axs[2])
axs[2].set_title("Number of Children") # Set boxplot title
axs[2].set_ylabel("Children Count") # Title Y axis label as children count

# Boxplot distribution for charges 
sns.boxplot(data=df_insurance, y="charges", ax=axs[3])
axs[3].set_title("Insurance Charges") # Set boxplot title
axs[3].set_ylabel("Charges") # Title Y axis label as charges

# Print grid visualising distribution counts
plt.show()
```

In hindsight, the outliers revealed in the boxplots for 'Insurance Charges' and 'BMI Distribution' could have been assessed during the ETL pipeline. Whilst a review of the statistical description for each column was made, the boxplots explicitly highlight outliers that could potentially impact the integrity of the analysis.

### **Correlation Analysis** 

A correlation analysis can reveal connections between different attributes present in the dataset. To visualise the relationship between smoking status, age and insurance charges, use Seaborn to create a scattergraph, using the categorical attribute as a <code>hue</code> to distinguish between smoker and non-smoker data.

<br/>

#### **Smoking Status, Age and Charges**

In the hypothesis, a question is raised about the correlation between age, smoking status and insurance charges. Seaborn provides a scatterplot with a <code>hue</code> parameter that can represent a categorical attribute alongside the numerical attributes of age and charges.

```python
# Create scatterplot with seaborn 
sns.scatterplot(
    data=df_insurance,
    x="age", # Asign X axis to age 
    y="charges", # Assign Y axis to charges
    hue="smoker" # Assign hue to smoker 
)

plt.title("Correlation between Age and Smoking Status on Charges") # Set scatterplot title 
plt.xlabel("Age") # Title X axis label as age
plt.ylabel("Insurance Charges") # Title Y axis label as insurance charges 

# Print scatterplot
plt.show()
```

The scatterplot shows a correlation between age, smoking status and insurance charges. Smokers are charged almost double the amount of non-smokers over a lifetime. However, it is important to acknowledge the uneven distribution of smokers in the dataset. Therefore, whilst this scatterplot is insightful, it should be interpreted with caution.

#### **Smokers Charges**

Display the DataFrame created in the ETL notebook to view the data. The DataFrame includes columns containing the minimum, mean and maximum insurance charges grouped by smoker status. Use these column names to plot an interactive bar chart and view the estimated charges based on smoker status.

```python
# Display smokers charges DataFrame
df_smokers_charges
```

To create an interactive bar chart with Plotly, assign the chart to the <code>smokers_charges_fig</code> variable. The <code>px.bar()</code> function uses the <code>df_smokers_charges</code> DataFrame, with <code>smoker</code> on the X-axis and the <code>min</code>, <code>mean</code> and <code>max</code> columns on the Y-axis.

Using <code>barmode="group"</code> displays the charge values side-by-side for smokers and non-smokers. Finally, <code>smokers_charges_fig.show()</code> displays the interactive bar chart.

```python
# Create a variable to plot bar chart 
smokers_charges_fig = px.bar(
    df_smokers_charges, # Use smokers charges DataFrame 
    x="smoker", # Assign X axis to smoker index 
    y=["min", "mean", "max"], # Assign Y axis to min, mean and max columns 
    barmode="group", # Group bars to display smokers vs non-smokers
    title="Insurance Charges by Smoking Status" # Set barchart title
)

# Display interactive barchart 
smokers_charges_fig.show()
```

The bar chart clearly displays higher healthcare insurance premiums for non-smokers.
