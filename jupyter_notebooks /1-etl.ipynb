#### **Aggregate Categorical Variable**

Create a DataFrame to apply predictive analysis on smokers status. Add columns including minimum, average and maximum healthcare insurance charges. 

```python
df_copy.head(4)
```

Create DataFrame featuring minimum, average and maximum charges between smokers and non-smoker. 

```python
# Creat DataFrame for smoking status charges
 
df_smoking_charges = df_copy.groupby('smoker')['charges'].agg(['min', 'mean', 'max'])
df_smoking_charges
```
